```python
# =======================================================
# STEP 4. OULAD verilənlər bazasında çatışmayan
# qiymətlərin (Missing Values) statistik təhlili
# =======================================================
#
# Məqsəd:
# OULAD verilənlər bazasına daxil olan bütün cədvəllərdə
# çatışmayan qiymətləri müəyyən etmək, onların sayını və
# faizini hesablamaq.
#
# Təhlil olunan göstəricilər:
# 1. Çatışmayan qiymətlərin sayı
# 2. Çatışmayan qiymətlərin faizi
# 3. Boş qiymətə malik atributların müəyyən edilməsi
#
# Nəticə:
# Çatışmayan qiymətlərin statistik cədvəli formalaşdırılır
# və sonrakı emal mərhələsi üçün ilkin baza yaradılır.
# =======================================================

datasets = {
    "studentInfo": studentInfo,
    "studentAssessment": studentAssessment,
    "studentRegistration": studentRegistration,
    "studentVle": studentVle,
    "assessments": assessments,
    "courses": courses,
    "vle": vle
}

missing_summary = []

print("=" * 70)
print("STEP 4. Çatışmayan qiymətlərin statistik təhlili")
print("=" * 70)

for name, df in datasets.items():

    print("\n" + "=" * 70)
    print(f"Dataset: {name}")
    print("=" * 70)

    total_rows = len(df)

    # Hər sütun üzrə boş qiymətlərin sayı
    missing = df.isnull().sum()

    # Yalnız boş qiyməti olan sütunlar
    missing = missing[missing > 0]

    if len(missing) == 0:

        print("Çatışmayan qiymət aşkar edilmədi.")

    else:

        result = pd.DataFrame({
            "Attribute": missing.index,
            "Missing Values": missing.values
        })

        result["Missing (%)"] = (
            result["Missing Values"] / total_rows * 100
        ).round(3)

        display(result)

        for _, row in result.iterrows():

            missing_summary.append({
                "Dataset": name,
                "Attribute": row["Attribute"],
                "Missing Values": int(row["Missing Values"]),
                "Missing (%)": row["Missing (%)"]
            })

print("\n" + "=" * 70)
print("OULAD verilənlər bazasında çatışmayan qiymətlərin ümumi statistikası")
print("=" * 70)

missing_summary = pd.DataFrame(missing_summary)

display(missing_summary)

print("\nÜmumi statistika")
print("-" * 40)

print(f"Dataset sayı               : {len(datasets)}")
print(f"Boş atributların sayı      : {len(missing_summary)}")
print(f"Ümumi Missing Values sayı  : {missing_summary['Missing Values'].sum():,}")

print("\n" + "=" * 70)
print("STEP 4 tamamlandı.")
print("Çatışmayan qiymətlərin statistik təhlili başa çatdı.")
print("=" * 70)
```

```python
# =======================================================
# STEP 4.1. Çatışmayan qiymətlərin vizuallaşdırılması
# =======================================================
#
# Məqsəd:
# OULAD verilənlər bazasında çatışmayan qiymətlərin
# paylanmasını vizual şəkildə göstərmək.
#
# Nəticə:
# Missing Values yalnız boş qiymət olan atributlar üzrə
# sütun diaqramı şəklində təqdim edilir.
# =======================================================

import matplotlib.pyplot as plt

datasets = {
    "studentInfo": studentInfo,
    "studentAssessment": studentAssessment,
    "studentRegistration": studentRegistration,
    "studentVle": studentVle,
    "assessments": assessments,
    "courses": courses,
    "vle": vle
}

missing_plot = []

for dataset_name, df in datasets.items():

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    for attribute, value in missing.items():

        missing_plot.append({
            "Dataset": dataset_name,
            "Attribute": attribute,
            "Missing": value
        })

missing_plot = pd.DataFrame(missing_plot)

plt.figure(figsize=(12,6))

labels = missing_plot["Dataset"] + "\n" + missing_plot["Attribute"]

plt.bar(labels,
        missing_plot["Missing"])

plt.title("OULAD verilənlər bazasında Missing Values")
plt.xlabel("Dataset / Atribut")
plt.ylabel("Missing Values")

plt.xticks(rotation=45, ha="right")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

plt.show()

print("="*70)
print("STEP 4.1 tamamlandı.")
print("Missing Values vizuallaşdırıldı.")
print("="*70)
```

```python
# =======================================================
# STEP 5. OULAD verilənlər bazasında təkrarlanan
# müşahidələrin (Duplicate Records) müəyyən edilməsi
# =======================================================
#
# Məqsəd:
# OULAD verilənlər bazasına daxil olan bütün əsas
# cədvəllərdə tam identik qeydləri müəyyən etmək və
# onların sayını hesablamaq.
#
# Təhlil olunan göstəricilər:
# 1. Ümumi müşahidələrin sayı
# 2. Təkrarlanan müşahidələrin sayı
#
# Nəticə:
# Təkrarlanan müşahidələrin statistik cədvəli
# formalaşdırılır.
# =======================================================

datasets = {
    "studentInfo": studentInfo,
    "studentAssessment": studentAssessment,
    "studentRegistration": studentRegistration,
    "studentVle": studentVle,
    "assessments": assessments,
    "courses": courses,
    "vle": vle
}

duplicate_summary = []

print("=" * 70)
print("STEP 5. Duplicate Records təhlili")
print("=" * 70)

for name, df in datasets.items():

    total = len(df)

    duplicates = df.duplicated().sum()

    duplicate_summary.append({

        "Dataset": name,

        "Total Records": total,

        "Duplicate Records": duplicates

    })

duplicate_summary = pd.DataFrame(duplicate_summary)

display(duplicate_summary)

print("\nÜmumi statistika")
print("-"*40)

print("Ümumi müşahidə sayı :", duplicate_summary["Total Records"].sum())

print("Ümumi duplicate sayı :", duplicate_summary["Duplicate Records"].sum())

print("="*70)
print("STEP 5 tamamlandı.")
print("Duplicate Records statistikası hazırlandı.")
print("="*70)
```

```python
# =======================================================
# STEP 6. Təkrarlanan müşahidələrin aradan qaldırılması
# və yekun yoxlanılması
# =======================================================
#
# Məqsəd:
# OULAD verilənlər bazasında müəyyən edilmiş tam
# təkrarlanan müşahidələri silmək və təmizlənmiş
# verilənlər bazasının struktur bütövlüyünü
# yenidən yoxlamaq.
#
# İcra olunan əməliyyatlar:
# 1. Tam identik qeydlərin silinməsi
# 2. Yenilənmiş müşahidə sayının hesablanması
# 3. Təkrar yoxlama aparılması
#
# Nəticə:
# Təkrarlanan müşahidələr verilənlər bazasından
# çıxarılır və yekun yoxlama nəticələri əldə edilir.
# =======================================================

datasets = {
    "studentInfo": studentInfo,
    "studentAssessment": studentAssessment,
    "studentRegistration": studentRegistration,
    "studentVle": studentVle,
    "assessments": assessments,
    "courses": courses,
    "vle": vle
}

clean_summary = []

print("=" * 70)
print("STEP 6. Duplicate Records-un aradan qaldırılması")
print("=" * 70)

for name, df in datasets.items():

    before_rows = len(df)

    duplicate_count = df.duplicated().sum()

    # Tam identik qeydlərin silinməsi
    df_clean = df.drop_duplicates()

    after_rows = len(df_clean)

    # Yenilənmiş DataFrame-i yaddaşa yaz
    datasets[name] = df_clean

    # Əsas dəyişənləri yenilə
    if name == "studentInfo":
        studentInfo = df_clean
    elif name == "studentAssessment":
        studentAssessment = df_clean
    elif name == "studentRegistration":
        studentRegistration = df_clean
    elif name == "studentVle":
        studentVle = df_clean
    elif name == "assessments":
        assessments = df_clean
    elif name == "courses":
        courses = df_clean
    elif name == "vle":
        vle = df_clean

    remaining_duplicates = df_clean.duplicated().sum()

    clean_summary.append({
        "Dataset": name,
        "Rows Before": before_rows,
        "Duplicates Removed": duplicate_count,
        "Rows After": after_rows,
        "Remaining Duplicates": remaining_duplicates
    })

clean_summary = pd.DataFrame(clean_summary)

print("\nTəmizləmə nəticələri")
print("-" * 70)

display(clean_summary)

print("\nYekun yoxlama")
print("-" * 70)

if clean_summary["Remaining Duplicates"].sum() == 0:
    print("✓ Bütün təkrarlanan müşahidələr uğurla aradan qaldırılmışdır.")
else:
    print("⚠ Bəzi cədvəllərdə hələ də duplicate qeydlər mövcuddur.")

print("\n" + "=" * 70)
print("STEP 6 tamamlandı.")
print("Verilənlər bazası təmizləndi və yekun yoxlama başa çatdı.")
print("=" * 70)
```

```python
# ==========================================================
# ƏLAVƏ A.6.1
# STEP 6.1. Təsviri statistik təhlil və Histogram analizi
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("="*70)
print("STEP 6.1")
print("Təsviri statistik təhlil və Histogram analizi")
print("="*70)

#----------------------------------------------------------
# Analiz ediləcək əsas ədədi atributlar
#----------------------------------------------------------

variables = [
    ("studentInfo", studentInfo, "studied_credits"),
    ("studentAssessment", studentAssessment, "score"),
    ("studentVle", studentVle, "sum_click"),
    ("studentRegistration", studentRegistration, "date_registration")
]

statistics = []

#----------------------------------------------------------
# Təsviri statistik göstəricilər
#----------------------------------------------------------

for dataset_name, df, column in variables:

    if column in df.columns:

        x = pd.to_numeric(df[column], errors="coerce").dropna()

        mean = x.mean()
        median = x.median()
        std = x.std()
        cv = (std / mean * 100) if mean != 0 else np.nan

        statistics.append({

            "Dataset": dataset_name,
            "Variable": column,
            "Count": len(x),
            "Mean": round(mean,2),
            "Median": round(median,2),
            "Std": round(std,2),
            "Min": round(x.min(),2),
            "Q1": round(x.quantile(0.25),2),
            "Q2": round(x.quantile(0.50),2),
            "Q3": round(x.quantile(0.75),2),
            "Max": round(x.max(),2),
            "CV (%)": round(cv,2)

        })

stats_df = pd.DataFrame(statistics)

print("\nTəsviri statistik göstəricilər")
display(stats_df)

#----------------------------------------------------------
# Histogram Analizi (Şəkil 3.8)
#----------------------------------------------------------

print("\nHistogramlar hazırlanır...")

for dataset_name, df, column in variables:

    if column in df.columns:

        x = pd.to_numeric(df[column], errors="coerce").dropna()

        plt.figure(figsize=(8,5))

        plt.hist(
            x,
            bins=30,
            edgecolor="black"
        )

        plt.title(f"{column} atributunun paylanması")
        plt.xlabel(column)
        plt.ylabel("Tezlik")
        plt.grid(alpha=0.3)

        plt.tight_layout()
        plt.show()

print("="*70)
print("STEP 6.1 tamamlandı.")
print("="*70)
```

```python
# ==========================================================
# ƏLAVƏ A.6.2
# STEP 6.2. BoxPlot analizi və kənar müşahidələrin qiymətləndirilməsi
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 6.2")
print("BoxPlot analizi və Outlier qiymətləndirilməsi")
print("=" * 70)

# ----------------------------------------------------------
# Analiz ediləcək əsas ədədi atributlar
# ----------------------------------------------------------

variables = [
    ("studentInfo", studentInfo, "studied_credits"),
    ("studentAssessment", studentAssessment, "score"),
    ("studentVle", studentVle, "sum_click"),
    ("studentRegistration", studentRegistration, "date_registration")
]

outlier_summary = []

# ----------------------------------------------------------
# BoxPlot və IQR analizi
# ----------------------------------------------------------

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, (dataset_name, df, column) in enumerate(variables):

    if column not in df.columns:
        continue

    x = pd.to_numeric(df[column], errors="coerce").dropna()

    # ---------- IQR hesablanması ----------

    Q1 = x.quantile(0.25)
    Q3 = x.quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = x[(x < lower) | (x > upper)]

    outlier_summary.append({

        "Dataset": dataset_name,
        "Variable": column,
        "Observations": len(x),
        "Q1": round(Q1,2),
        "Q3": round(Q3,2),
        "IQR": round(IQR,2),
        "Lower Limit": round(lower,2),
        "Upper Limit": round(upper,2),
        "Outliers": len(outliers),
        "Outlier (%)": round(len(outliers)/len(x)*100,3)

    })

    # ---------- BoxPlot ----------

    axes[i].boxplot(
        x,
        vert=True,
        patch_artist=True
    )

    axes[i].set_title(column)
    axes[i].set_ylabel("Qiymət")
    axes[i].grid(alpha=0.3)

plt.suptitle("Şəkil 3.9. Əsas atributlar üzrə BoxPlot analizi",
             fontsize=14)

plt.tight_layout()
plt.show()

# ----------------------------------------------------------
# Nəticələrin cədvəli
# ----------------------------------------------------------

outlier_df = pd.DataFrame(outlier_summary)

print("\nOutlier Statistikası")
display(outlier_df)

print("=" * 70)
print("STEP 6.2 tamamlandı.")
print("=" * 70)
```

```python
# ==========================================================
# STEP 7
# Assessment nəticələrinin tələbə səviyyəsində aqreqasiyası
# ==========================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 7")
print("Qiymətləndirmə nəticələrinin aqreqasiya edilməsi")
print("=" * 70)


#----------------------------------------------------------
# 1. studentAssessment cədvəlinin yoxlanılması
#----------------------------------------------------------

print("\nstudentAssessment ölçüsü:")
print(studentAssessment.shape)


#----------------------------------------------------------
# 2. Score sütununda boş qiymətlərin çıxarılması
#----------------------------------------------------------

assessment = studentAssessment.dropna(subset=["score"]).copy()

print("\nBoş score qiymətləri çıxarıldı.")

print("Yeni ölçü:", assessment.shape)


#----------------------------------------------------------
# 3. Tələbə üzrə aqreqasiya
#----------------------------------------------------------

assessment_summary = (
    assessment
    .groupby("id_student")
    .agg(
        Average_Score=("score","mean"),
        Maximum_Score=("score","max"),
        Minimum_Score=("score","min"),
        Assessment_Count=("score","count")
    )
    .reset_index()
)

print("\nAssessment Summary yaradıldı.")


#----------------------------------------------------------
# 4. Rəqəmlərin yuvarlaqlaşdırılması
#----------------------------------------------------------

assessment_summary["Average_Score"] = \
assessment_summary["Average_Score"].round(2)

assessment_summary["Maximum_Score"] = \
assessment_summary["Maximum_Score"].round(2)

assessment_summary["Minimum_Score"] = \
assessment_summary["Minimum_Score"].round(2)


#----------------------------------------------------------
# 5. İlk müşahidələr (Cədvəl 3.10)
#----------------------------------------------------------

print("\nİlk 10 müşahidə")
print(assessment_summary.head(10))


#----------------------------------------------------------
# 6. Aqreqasiya statistikası (Cədvəl 3.11)
#----------------------------------------------------------

summary_table = pd.DataFrame({

    "Göstərici":[
        "İlkin qiymətləndirmə qeydlərinin sayı",
        "Aqreqasiya olunmuş tələbə sayı",
        "Yaradılmış xüsusiyyətlərin sayı",
        "Yeni yaradılmış cədvəl"
    ],

    "Qiymət":[
        len(studentAssessment),
        assessment_summary.shape[0],
        assessment_summary.shape[1]-1,
        "assessment_summary"
    ]

})

print("\n")
print("="*60)
print("Cədvəl 3.11")
print("="*60)

print(summary_table)


#----------------------------------------------------------
# 7. Təsviri statistika
#----------------------------------------------------------

statistics = assessment_summary.describe().T

statistics = statistics[[
    "count",
    "mean",
    "std",
    "min",
    "25%",
    "50%",
    "75%",
    "max"
]]

statistics = statistics.round(2)

print("\n")
print("="*60)
print("Assessment Summary Statistikası")
print("="*60)

print(statistics)


#----------------------------------------------------------
# 8. Əsas orta qiymətlər
#----------------------------------------------------------

print("\n")
print("="*60)
print("Əsas statistik göstəricilər")
print("="*60)

print(f"Average Score      : {assessment_summary['Average_Score'].mean():.2f}")

print(f"Maximum Score      : {assessment_summary['Maximum_Score'].mean():.2f}")

print(f"Minimum Score      : {assessment_summary['Minimum_Score'].mean():.2f}")

print(f"Assessment Count   : {assessment_summary['Assessment_Count'].mean():.2f}")


#----------------------------------------------------------
# 9. CSV faylına yazılması
#----------------------------------------------------------

assessment_summary.to_csv(
    "assessment_summary.csv",
    index=False
)

print("\nassessment_summary.csv faylı yaradıldı.")


#----------------------------------------------------------
# 10. Son məlumat
#----------------------------------------------------------

print("\n")
print("="*70)
print("STEP 7 tamamlandı.")
print("="*70)
```

```python
# ==========================================================
# STEP 8.1
# assessment_summary Integration
# Quality Evaluation Dataset (QED) yaradılması
# ==========================================================

import pandas as pd
import matplotlib.pyplot as plt

print("=" * 70)
print("STEP 8.1")
print("assessment_summary inteqrasiyası")
print("=" * 70)

# ----------------------------------------------------------
# 1. Faylların yüklənməsi
# ----------------------------------------------------------

merged_student = pd.read_csv("merged_student.csv")
assessment_summary = pd.read_csv("assessment_summary.csv")

print("\nÖlçülər")
print("merged_student     :", merged_student.shape)
print("assessment_summary :", assessment_summary.shape)

# ----------------------------------------------------------
# 2. LEFT MERGE
# ----------------------------------------------------------

QED = pd.merge(
    merged_student,
    assessment_summary,
    how="left",
    on="id_student"
)

print("\nİnteqrasiya tamamlandı.")
print("QED ölçüsü:")
print(QED.shape)

print("\nİlk 10 müşahidə")
print(QED.head(10))

# ----------------------------------------------------------
# 3. Boş qiymətlərin statistikası
# ----------------------------------------------------------

missing = QED[
    [
        "Average_Score",
        "Maximum_Score",
        "Minimum_Score",
        "Assessment_Count"
    ]
].isnull().sum()

print("\nBoş qiymətlər")
print(missing)

# ----------------------------------------------------------
# 4. Assessment məlumatının mövcudluğu
# ----------------------------------------------------------

students_with_scores = QED["Average_Score"].notna().sum()
students_without_scores = QED["Average_Score"].isna().sum()

print("\nAssessment məlumatı olan tələbələr :", students_with_scores)
print("Assessment məlumatı olmayan tələbələr:", students_without_scores)

# ----------------------------------------------------------
# 5. Qrafik
# ----------------------------------------------------------

plt.figure(figsize=(8,6))

bars = plt.bar(
    ["Assessment məlumatı var",
     "Assessment məlumatı yoxdur"],
    [students_with_scores,
     students_without_scores]
)

plt.title(
    "QED verilənlər bazasında qiymətləndirmə məlumatlarının mövcudluğu",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel("Tələbələrin sayı")
plt.grid(axis="y", linestyle="--", alpha=0.4)

# Sütunların üzərinə qiymətləri yaz

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 150,
        f"{int(height):,}",
        ha="center",
        fontsize=11,
        fontweight="bold"
    )

plt.tight_layout()

plt.savefig(
    "Figure_3_11_QED_Assessment_Coverage.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nŞəkil Figure_3_11_QED_Assessment_Coverage.png faylı kimi saxlanıldı.")

# ----------------------------------------------------------
# 6. QED faylının saxlanılması
# ----------------------------------------------------------

QED.to_csv(
    "QED.csv",
    index=False
)

print("\nQED.csv yaradıldı.")

print("\n" + "=" * 70)
print("STEP 8.1 tamamlandı.")
print("=" * 70)
```
